```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_absolute_error, r2_score
import joblib

# 1. LOAD THE CLEAN DATA
# We don't need to clean again. We just load the result from File 1.
df = pd.read_csv('Algerian_forest_fires_cleaned_dataset.csv')

# Drop Date columns (Day, Month, Year) as they are not needed for prediction
df = df.drop(['day', 'month', 'year'], axis=1)

# 2. PREPROCESSING
# Separate Independent (X) and Dependent (y) variables
# We drop 'FWI' because it is the target.
# We drop 'Classes' because it is a "Cheat Sheet" (Label Leakage).
X = df.drop(['FWI', 'Classes'], axis=1)
y = df['FWI']

# Train-Test Split (75% Study, 25% Exam)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Feature Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. MODEL TRAINING & COMPARISON

# --- Model 1: Linear Regression ---
lin_reg = LinearRegression()
lin_reg.fit(X_train_scaled, y_train)
y_pred_lin = lin_reg.predict(X_test_scaled)
print("Linear Regression R2:", r2_score(y_test, y_pred_lin))

# --- Model 2: Ridge Regression (The chosen one for the project) ---
ridge_reg = Ridge(alpha=10.0)
ridge_reg.fit(X_train_scaled, y_train)
y_pred_ridge = ridge_reg.predict(X_test_scaled)
print("Ridge Regression R2:", r2_score(y_test, y_pred_ridge))

# --- Model 3: Lasso Regression ---
lasso_reg = Lasso(alpha=0.2)
lasso_reg.fit(X_train_scaled, y_train)
y_pred_lasso = lasso_reg.predict(X_test_scaled)
print("Lasso Regression R2:", r2_score(y_test, y_pred_lasso))

# --- Model 4: Elastic Net ---
elastic_reg = ElasticNet(alpha=0.2, l1_ratio=0.5)
elastic_reg.fit(X_train_scaled, y_train)
y_pred_elastic = elastic_reg.predict(X_test_scaled)
print("Elastic Net R2:", r2_score(y_test, y_pred_elastic))


# 4. SAVING THE MODEL (PICKLING)
# We choose Ridge Regression because it handles multicollinearity well.
# We use 'joblib' to save the files.

joblib.dump(scaler, 'scaler.pkl')
joblib.dump(ridge_reg, 'ridge.pkl')

print("\nSUCCESS: Models trained. 'scaler.pkl' and 'ridge.pkl' saved to your folder.")
```
